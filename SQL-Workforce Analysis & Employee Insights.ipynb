```python
# Inspect the table structure

query = """
PRAGMA table_info(employees);
"""

pd.read_sql_query(query, conn)
```

```python
# Check for duplicate employees

query = """
SELECT 
    Employee_ID,
    COUNT(*) AS record_count
FROM employees
GROUP BY Employee_ID
HAVING COUNT(*) > 1;
"""

pd.read_sql_query(query, conn)
```

```python
# Check for missing values

query = """
SELECT
    SUM(CASE WHEN Employee_ID IS NULL THEN 1 ELSE 0 END) AS missing_employee_id,
    SUM(CASE WHEN Gender IS NULL THEN 1 ELSE 0 END) AS missing_gender,
    SUM(CASE WHEN Age IS NULL THEN 1 ELSE 0 END) AS missing_age,
    SUM(CASE WHEN Department IS NULL THEN 1 ELSE 0 END) AS missing_department,
    SUM(CASE WHEN Salary IS NULL THEN 1 ELSE 0 END) AS missing_salary,
    SUM(CASE WHEN Performance_Rating IS NULL THEN 1 ELSE 0 END) AS missing_performance,
    SUM(CASE WHEN Attrition IS NULL THEN 1 ELSE 0 END) AS missing_attrition,
    SUM(CASE WHEN Satisfaction_Score IS NULL THEN 1 ELSE 0 END) AS missing_satisfaction
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
# Employees by Gender

query = """
SELECT
    Gender,
    COUNT(*) AS employee_count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM employees), 2) AS percentage
FROM employees
GROUP BY Gender
ORDER BY employee_count DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Employees by Department

query = """
SELECT
    Department,
    COUNT(*) AS employee_count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM employees), 2) AS workforce_percentage
FROM employees
GROUP BY Department
ORDER BY employee_count DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Employees by employment type

query = """SELECT
    Employment_Type,
    COUNT(*) AS employee_count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM employees), 2) AS workforce_percentage,
    ROUND(AVG(Salary), 2) AS average_salary
FROM employees
GROUP BY Employment_Type
ORDER BY employee_count DESC; """

pd.read_sql_query(query, conn)
```

```python
# Employees by work mode

query = """
SELECT
    Work_Mode,
    COUNT(*) AS employee_count,
    ROUND(AVG(Satisfaction_Score), 2) AS avg_satisfaction
FROM employees
GROUP BY Work_Mode
ORDER BY employee_count DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Overall salary statistics

query = """
SELECT
    COUNT(*) AS employees,
    ROUND(MIN(Salary), 2) AS minimum_salary,
    ROUND(AVG(Salary), 2) AS average_salary,
    ROUND(MAX(Salary), 2) AS maximum_salary,
    ROUND(AVG(Bonus), 2) AS average_bonus
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
# Salary by Department

query = """
SELECT
    Department,
    COUNT(*) AS employee_count,
    ROUND(AVG(Salary), 2) AS average_salary,
    ROUND(MIN(Salary), 2) AS minimum_salary,
    ROUND(MAX(Salary), 2) AS maximum_salary,
    ROUND(AVG(Bonus), 2) AS average_bonus
FROM employees
GROUP BY Department
ORDER BY average_salary DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Employees with the highest total compensation (Salary + Bonus)

query = """
SELECT
    Employee_ID,
    First_Name || ' ' || Last_Name AS Employee_Name,
    Department,
    Job_Title,
    Salary,
    Bonus,
    Salary + Bonus AS Total_Compensation
FROM employees
ORDER BY Total_Compensation DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
# Salary ranking within departments

query = """
WITH ranked_employees AS (
    SELECT
        Employee_ID,
        First_Name || ' ' || Last_Name AS Employee_Name,
        Department,
        Job_Title,
        Salary,
        RANK() OVER (
            PARTITION BY Department
            ORDER BY Salary DESC
        ) AS department_salary_rank
    FROM employees
)
SELECT *
FROM ranked_employees
WHERE department_salary_rank <= 3
ORDER BY Department, department_salary_rank;
"""

pd.read_sql_query(query, conn)
```

```python
# Performance distribution

query = """
SELECT
    Performance_Rating,
    COUNT(*) AS employee_count,
    ROUND(
        COUNT(*) * 100.0 / (SELECT COUNT(*) FROM employees),
        2
    ) AS percentage
FROM employees
GROUP BY Performance_Rating
ORDER BY Performance_Rating DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Performance by Department

query = """
SELECT
    Department,
    COUNT(*) AS employee_count,
    ROUND(AVG(Performance_Rating), 2) AS avg_performance,
    ROUND(AVG(Salary), 2) AS avg_salary,
    ROUND(AVG(Satisfaction_Score), 2) AS avg_satisfaction
FROM employees
GROUP BY Department
ORDER BY avg_performance DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# High Performers

query = """
SELECT
    Employee_ID,
    First_Name || ' ' || Last_Name AS Employee_Name,
    Department,
    Job_Title,
    Performance_Rating,
    Salary,
    Satisfaction_Score,
    Promotion
FROM employees
WHERE Performance_Rating >= 4
ORDER BY Performance_Rating DESC, Salary DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Satisfaction by Department

query = """
SELECT
    Department,
    COUNT(*) AS employee_count,
    ROUND(AVG(Satisfaction_Score), 2) AS avg_satisfaction,
    MIN(Satisfaction_Score) AS lowest_score,
    MAX(Satisfaction_Score) AS highest_score
FROM employees
GROUP BY Department
ORDER BY avg_satisfaction DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Satisfaction and Work-mode

query = """
SELECT
    Work_Mode,
    COUNT(*) AS employee_count,
    ROUND(AVG(Satisfaction_Score), 2) AS avg_satisfaction,
    ROUND(AVG(Overtime_Hours), 2) AS avg_overtime,
    ROUND(AVG(Leave_Days), 2) AS avg_leave_days
FROM employees
GROUP BY Work_Mode
ORDER BY avg_satisfaction DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Employees with high overtime

query = """
SELECT
    Employee_ID,
    First_Name || ' ' || Last_Name AS Employee_Name,
    Department,
    Job_Title,
    Overtime_Hours,
    Satisfaction_Score,
    Attrition
FROM employees
WHERE Overtime_Hours >= 50
ORDER BY Overtime_Hours DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Overtime by Department

query = """
SELECT
    Department,
    COUNT(*) AS employee_count,
    ROUND(AVG(Overtime_Hours), 2) AS avg_overtime,
    ROUND(MAX(Overtime_Hours), 2) AS max_overtime,
    ROUND(AVG(Satisfaction_Score), 2) AS avg_satisfaction
FROM employees
GROUP BY Department
ORDER BY avg_overtime DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Overall Attrition Rate

query = """
SELECT
    COUNT(*) AS total_employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
    SUM(CASE WHEN Attrition = 'No' THEN 1 ELSE 0 END) AS employees_retained,
    ROUND(
        SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS attrition_rate
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
# Attrition by Department

query = """
SELECT
    Department,
    COUNT(*) AS total_employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
    ROUND(
        SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS attrition_rate
FROM employees
GROUP BY Department
ORDER BY attrition_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Attrition by Gender

query = """
SELECT
    Gender,
    COUNT(*) AS total_employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
    ROUND(
        SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS attrition_rate
FROM employees
GROUP BY Gender
ORDER BY attrition_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Attrition by Gender and Department 

query = """
SELECT
    Gender,
    Department,
    COUNT(*) AS total_employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
    ROUND(
        SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS attrition_rate
FROM employees
GROUP BY Gender, Department
ORDER BY Gender, attrition_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Attrition by Employment type

query = """
SELECT
    Employment_Type,
    COUNT(*) AS total_employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
    ROUND(
        SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS attrition_rate,
    ROUND(AVG(Salary), 2) AS avg_salary
FROM employees
GROUP BY Employment_Type
ORDER BY attrition_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Compare satisfaction, overtime, salary, and performance between employees who left and stayed

query = """
SELECT
    Attrition,
    COUNT(*) AS employee_count,
    ROUND(AVG(Satisfaction_Score), 2) AS avg_satisfaction,
    ROUND(AVG(Overtime_Hours), 2) AS avg_overtime,
    ROUND(AVG(Salary), 2) AS avg_salary,
    ROUND(AVG(Performance_Rating), 2) AS avg_performance
FROM employees
GROUP BY Attrition;
"""

pd.read_sql_query(query, conn)
```

```python
# Employees meeting potential attrition risk criteria


query = """
SELECT
    Employee_ID,
    First_Name || ' ' || Last_Name AS Employee_Name,
    Department,
    Job_Title,
    Salary,
    Overtime_Hours,
    Satisfaction_Score,
    Performance_Rating,
    Promotion
FROM employees
WHERE Satisfaction_Score <= 4
  AND Overtime_Hours >= 40
ORDER BY Satisfaction_Score ASC, Overtime_Hours DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Promotion rate

query = """
SELECT
    COUNT(*) AS total_employees,
    SUM(CASE WHEN Promotion = 'Yes' THEN 1 ELSE 0 END) AS promoted_employees,
    ROUND(
        SUM(CASE WHEN Promotion = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS promotion_rate
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
# Promotion by Department

query = """
SELECT
    Department,
    COUNT(*) AS total_employees,
    SUM(CASE WHEN Promotion = 'Yes' THEN 1 ELSE 0 END) AS promoted,
    ROUND(
        SUM(CASE WHEN Promotion = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS promotion_rate,
    ROUND(AVG(Performance_Rating), 2) AS avg_performance
FROM employees
GROUP BY Department
ORDER BY promotion_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# High performers who have not been promoted

query = """
SELECT
    Employee_ID,
    First_Name || ' ' || Last_Name AS Employee_Name,
    Department,
    Job_Title,
    Performance_Rating,
    Experience_Years,
    Salary,
    Satisfaction_Score
FROM employees
WHERE Performance_Rating >= 4
  AND Promotion = 'No'
ORDER BY Performance_Rating DESC, Experience_Years DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Training hours by Department

query = """
SELECT
    Department,
    COUNT(*) AS employee_count,
    ROUND(AVG(Training_Hours), 2) AS avg_training_hours,
    ROUND(AVG(Performance_Rating), 2) AS avg_performance,
    ROUND(AVG(Satisfaction_Score), 2) AS avg_satisfaction
FROM employees
GROUP BY Department
ORDER BY avg_training_hours DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Training and Performance

query = """
SELECT
    CASE
        WHEN Training_Hours < 25 THEN 'Low Training'
        WHEN Training_Hours BETWEEN 25 AND 50 THEN 'Medium Training'
        ELSE 'High Training'
    END AS training_group,
    COUNT(*) AS employee_count,
    ROUND(AVG(Performance_Rating), 2) AS avg_performance,
    ROUND(AVG(Satisfaction_Score), 2) AS avg_satisfaction
FROM employees
GROUP BY training_group
ORDER BY avg_performance DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Experience Groups

query = """
SELECT
    CASE
        WHEN Experience_Years < 3 THEN '0-2 Years'
        WHEN Experience_Years BETWEEN 3 AND 7 THEN '3-7 Years'
        WHEN Experience_Years BETWEEN 8 AND 15 THEN '8-15 Years'
        ELSE '16+ Years'
    END AS experience_group,
    COUNT(*) AS employee_count,
    ROUND(AVG(Salary), 2) AS avg_salary,
    ROUND(AVG(Performance_Rating), 2) AS avg_performance,
    ROUND(AVG(Satisfaction_Score), 2) AS avg_satisfaction,
    ROUND(
        SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS attrition_rate
FROM employees
GROUP BY experience_group
ORDER BY 
    CASE experience_group
        WHEN '0-2 Years' THEN 1
        WHEN '3-7 Years' THEN 2
        WHEN '8-15 Years' THEN 3
        ELSE 4
    END;
"""

pd.read_sql_query(query, conn)
```

```python
# Job Titles with the highest salaries

query = """
SELECT
    Job_Title,
    COUNT(*) AS employee_count,
    ROUND(AVG(Salary), 2) AS avg_salary,
    ROUND(AVG(Bonus), 2) AS avg_bonus,
    ROUND(AVG(Performance_Rating), 2) AS avg_performance
FROM employees
GROUP BY Job_Title
HAVING COUNT(*) >= 5
ORDER BY avg_salary DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Manager-level workforce and attrition analysis

query = """
SELECT
    Manager,
    COUNT(*) AS team_size,
    ROUND(AVG(Salary), 2) AS avg_salary,
    ROUND(AVG(Performance_Rating), 2) AS avg_performance,
    ROUND(AVG(Satisfaction_Score), 2) AS avg_satisfaction,
    ROUND(AVG(Overtime_Hours), 2) AS avg_overtime,
    ROUND(
        SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS attrition_rate
FROM employees
GROUP BY Manager
HAVING COUNT(*) >= 5
ORDER BY attrition_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# State-level workforce, salary, satisfaction, and attrition analysis

query = """
SELECT
    State,
    COUNT(*) AS employee_count,
    ROUND(AVG(Salary), 2) AS avg_salary,
    ROUND(AVG(Satisfaction_Score), 2) AS avg_satisfaction,
    ROUND(
        SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS attrition_rate
FROM employees
GROUP BY State
ORDER BY employee_count DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Compensation efficiency: Provides a simple comparison of average salary relative to average performance.

query = """
SELECT
    Department,
    ROUND(AVG(Salary), 2) AS avg_salary,
    ROUND(AVG(Performance_Rating), 2) AS avg_performance,
    ROUND(
        AVG(Salary) / AVG(Performance_Rating),
        2
    ) AS salary_per_performance_point
FROM employees
GROUP BY Department
ORDER BY salary_per_performance_point DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Employee Segmentation

query = """
SELECT
    CASE
        WHEN Performance_Rating >= 4
             AND Satisfaction_Score >= 7
             THEN 'High Performer - Highly Satisfied'

        WHEN Performance_Rating >= 4
             AND Satisfaction_Score < 7
             THEN 'High Performer - Low Satisfaction'

        WHEN Performance_Rating < 4
             AND Satisfaction_Score >= 7
             THEN 'Lower Performer - Highly Satisfied'

        ELSE 'Lower Performer - Low Satisfaction'
    END AS employee_segment,

    COUNT(*) AS employee_count,

    ROUND(AVG(Salary), 2) AS avg_salary,
    ROUND(AVG(Overtime_Hours), 2) AS avg_overtime,
    ROUND(AVG(Training_Hours), 2) AS avg_training,
    ROUND(
        SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS attrition_rate

FROM employees
GROUP BY employee_segment
ORDER BY employee_count DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Overall HR snapshot 

query = """
SELECT
    COUNT(*) AS total_employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
    ROUND(
        SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS attrition_rate,
    ROUND(AVG(Salary), 2) AS avg_salary,
    ROUND(AVG(Bonus), 2) AS avg_bonus,
    ROUND(AVG(Performance_Rating), 2) AS avg_performance,
    ROUND(AVG(Satisfaction_Score), 2) AS avg_satisfaction,
    ROUND(AVG(Overtime_Hours), 2) AS avg_overtime,
    ROUND(AVG(Training_Hours), 2) AS avg_training_hours,
    SUM(CASE WHEN Promotion = 'Yes' THEN 1 ELSE 0 END) AS employees_promoted,
    ROUND(
        SUM(CASE WHEN Promotion = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS promotion_rate
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
# Department Performance Scorecard

squery = """
WITH department_metrics AS (
    SELECT
        Department,
        COUNT(*) AS employee_count,

        AVG(Salary) AS avg_salary,
        AVG(Performance_Rating) AS avg_performance,
        AVG(Satisfaction_Score) AS avg_satisfaction,
        AVG(Overtime_Hours) AS avg_overtime,
        AVG(Training_Hours) AS avg_training,

        SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) * 100.0
            / COUNT(*) AS attrition_rate,

        SUM(CASE WHEN Promotion = 'Yes' THEN 1 ELSE 0 END) * 100.0
            / COUNT(*) AS promotion_rate

    FROM employees
    GROUP BY Department
),

normalized_metrics AS (
    SELECT
        *,
        
        -- Normalize performance from 1-5 to 0-100
        ((avg_performance - 1) / 4.0) * 100 AS performance_score,

        -- Normalize satisfaction from 1-10 to 0-100
        ((avg_satisfaction - 1) / 9.0) * 100 AS satisfaction_score,

        -- Promotion rate already expressed as a percentage
        promotion_rate AS promotion_score,

        -- Higher training is treated as positive
        -- relative to the highest department training level
        (avg_training * 100.0 /
            MAX(avg_training) OVER ()) AS training_score,

        -- Lower attrition is better
        (1 - attrition_rate /
            MAX(attrition_rate) OVER ()) * 100 AS retention_score

    FROM department_metrics
)

SELECT
    Department,
    employee_count,

    ROUND(avg_salary, 2) AS avg_salary,
    ROUND(avg_performance, 2) AS avg_performance,
    ROUND(avg_satisfaction, 2) AS avg_satisfaction,
    ROUND(avg_overtime, 2) AS avg_overtime,
    ROUND(avg_training, 2) AS avg_training,
    ROUND(attrition_rate, 2) AS attrition_rate,
    ROUND(promotion_rate, 2) AS promotion_rate,

    ROUND(
        performance_score * 0.30
        + satisfaction_score * 0.25
        + promotion_score * 0.15
        + training_score * 0.10
        + retention_score * 0.20,
        2
    ) AS department_score

FROM normalized_metrics

ORDER BY department_score DESC;
"""

pd.read_sql_query(squery, conn)
```
